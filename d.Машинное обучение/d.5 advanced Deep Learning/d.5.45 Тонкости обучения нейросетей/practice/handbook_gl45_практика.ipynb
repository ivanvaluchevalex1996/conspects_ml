```python
import torch
import torch.nn as nn
```

---
# §45.1 · Kaiming



```python
layer = nn.Linear(256, 128)
nn.init.kaiming_normal_(layer.weight, nonlinearity='relu')
nn.init.zeros_(layer.bias)
print('std He', round(layer.weight.std().item(), 3), 'target', round((2/256)**0.5, 3))
```

---
# §45.2 · BatchNorm1d vs LayerNorm



```python
x = torch.randn(16, 64)
bn = nn.BatchNorm1d(64)(x)
ln = nn.LayerNorm(64)(x)
print('LN mean', round(ln.mean().item(), 3), 'std', round(ln.std().item(), 2))
```

---
# §45.3 · clip grad



```python
model = nn.Sequential(nn.Linear(64, 64), nn.ReLU(), nn.Linear(64, 10))
loss = model(torch.randn(8, 64)).sum()
loss.backward()
norm = torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
print('grad norm was', round(norm.item(), 2))
```
