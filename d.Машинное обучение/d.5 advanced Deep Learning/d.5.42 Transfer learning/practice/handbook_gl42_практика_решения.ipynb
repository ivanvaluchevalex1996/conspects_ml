# §42.2 · заморозка и замена головы



```python
pretrained = nn.Sequential(
    nn.Linear(64, 128), nn.ReLU(),
    nn.Linear(128, 64), nn.ReLU(),
    nn.Linear(64, 1000),
)
for p in pretrained.parameters():
    p.requires_grad = False
pretrained[-1] = nn.Linear(64, 5)
trainable = [n for n, p in pretrained.named_parameters() if p.requires_grad]
print('trainable:', trainable)
opt = torch.optim.Adam((p for p in pretrained.parameters() if p.requires_grad), lr=1e-3)
```
