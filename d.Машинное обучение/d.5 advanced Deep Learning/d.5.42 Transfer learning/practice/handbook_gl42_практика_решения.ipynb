# §42.2 · Заморозка + голова

## Зачем этот фрагмент

Feature extraction: тело не учится (`requires_grad=False`), учится только новая голова под твои классы.

См. [d.5.42 §3](../d.5.42 Transfer learning.md#3-два-режима-handbook-422).

## Что проверить

В `named_parameters()` trainable только веса последнего `Linear`.


```python
pretrained = nn.Sequential(
    nn.Linear(64, 128), nn.ReLU(),
    nn.Linear(128, 64), nn.ReLU(),
    nn.Linear(64, 1000),
)
for p in pretrained.parameters():
    p.requires_grad = False
pretrained[-1] = nn.Linear(64, 5)
print('trainable:', [n for n, p in pretrained.named_parameters() if p.requires_grad])
```
