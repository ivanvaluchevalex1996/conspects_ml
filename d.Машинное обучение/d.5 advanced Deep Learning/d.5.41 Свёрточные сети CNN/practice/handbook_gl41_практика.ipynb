# §41.2 · Conv2d

## Зачем этот фрагмент

Закрепляешь размерность **NCHW** и рост числа каналов без смены H×W при `padding=1`, kernel=3.

См. [d.5.41 §3](../d.5.41 Свёрточные сети CNN.md#3-каналы-и-карты-признаков-handbook-412).

## Что проверить

in **(1, 3, 32, 32)** → out **(1, 16, 32, 32)**.


```python
conv = nn.Conv2d(3, 16, kernel_size=3, padding=1)
img = torch.randn(1, 3, 32, 32)
feat = conv(img)
print('in:', tuple(img.shape), '→ out:', tuple(feat.shape))
```

# §41.3 · padding / stride

## Зачем этот фрагмент

Видишь на цифрах формулу «размер карты»: padding и stride — главный рычаг глубины vs разрешения.

См. [d.5.41 §4](../d.5.41 Свёрточные сети CNN.md#4-padding-и-stride-handbook-413).

## Что проверить

`padding=0` → карта меньше; `stride=2` → H,W делятся на 2 (при pad=1, k=3).


```python
x = torch.randn(1, 1, 28, 28)
for p, s in [(0, 1), (1, 1), (1, 2)]:
    y = nn.Conv2d(1, 1, kernel_size=3, padding=p, stride=s)(x)
    print(f'padding={p}, stride={s}:', tuple(y.shape[2:]))
```

# §41.5 · Conv vs Linear

## Зачем этот фрагмент

Параметры Conv **не растут** с площадью картинки — sharing весов по пространству.

См. [d.5.41 §6](../d.5.41 Свёрточные сети CNN.md#6-из-чего-собрана-cnn-handbook-415).

Дополнительно: [deep_ml](./deep_ml/README.md).

## Что проверить

Conv2d параметров на порядки меньше, чем `Linear(32*32*3 → 16)`.


```python
conv = nn.Conv2d(3, 16, kernel_size=3, padding=1)
fc = nn.Linear(32 * 32 * 3, 16)
print('Conv2d:', sum(p.numel() for p in conv.parameters()))
print('Linear:', sum(p.numel() for p in fc.parameters()))
```
