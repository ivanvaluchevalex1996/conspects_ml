```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

---
# §41.2 · Conv2d



```python
conv = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
img = torch.randn(1, 3, 32, 32)
feat = conv(img)
print('in', tuple(img.shape), 'out', tuple(feat.shape))
```

---
# §41.3 · padding / stride



```python
x = torch.randn(1, 1, 28, 28)
for p, s in [(0, 1), (1, 1), (1, 2)]:
    y = nn.Conv2d(1, 1, kernel_size=3, padding=p, stride=s)(x)
    print(f'padding={p}, stride={s}:', tuple(y.shape[2:]))
```

---
# §41.5 · параметры Conv2d vs Linear



```python
conv = nn.Conv2d(3, 16, kernel_size=3, padding=1)
fc = nn.Linear(32 * 32 * 3, 16)
print('Conv2d', sum(p.numel() for p in conv.parameters()))
print('Linear', sum(p.numel() for p in fc.parameters()))
```
