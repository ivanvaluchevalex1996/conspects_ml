## Зачем эта задача

Один нейрон с сигмоидой и MSE: **полный цикл** update весов вручную — мост к гл. 35.


## Условие

`train_neuron(...)` → w, b, список MSE по эпохам; эталонные числа из #25.


```python
import numpy as np


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"


def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))


def train_neuron(features, labels, initial_weights, initial_bias, learning_rate, epochs):
    w = initial_weights.astype(float).copy()
    b = float(initial_bias)
    mses = []
    n = len(labels)
    for _ in range(epochs):
        z = features @ w + b
        a = sigmoid(z)
        mses.append(round(float(np.mean((a - labels) ** 2)), 4))
        da = 2 * (a - labels) / n
        dz = da * a * (1 - a)
        w -= learning_rate * (features.T @ dz)
        b -= learning_rate * np.sum(dz)
    return w, b, mses


features = np.array([[1.0, 2.0], [2.0, 1.0], [-1.0, -2.0]])
labels = np.array([1, 0, 0], dtype=float)
uw, ub, mses = train_neuron(features, labels, np.array([0.1, -0.2]), 0.0, 0.1, 2)
need(np.allclose(uw, [0.1036, -0.1425], atol=1e-4))
need(abs(ub - (-0.0167)) < 1e-4 and mses == [0.3033, 0.2942])
print("weights", uw, "bias", ub, "mse", mses)
```

## Построчно

- forward: z, σ, MSE.
- chain: dL/da → dz → обновление w,b.
- `features.T @ dz` — градиент по весам батча.
