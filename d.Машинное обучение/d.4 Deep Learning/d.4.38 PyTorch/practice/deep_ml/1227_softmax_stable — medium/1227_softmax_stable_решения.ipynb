# Deep-ML #1227 · Numerically Stable Softmax · **medium**

Реализуй `stable_softmax(x, dim=-1)`: вычти максимум по `dim` перед `exp`, чтобы не было переполнения. Результат должен совпадать с `torch.softmax`.

[#1227](https://www.deep-ml.com/problems/1227) · [README.md](./README.md) · эталон [1227_softmax_stable_решения.ipynb](./1227_softmax_stable_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def stable_softmax(x, dim=-1):
    z = x - x.max(dim=dim, keepdim=True).values
    e = torch.exp(z)
    return e / e.sum(dim=dim, keepdim=True)
```

```python
need(close(stable_softmax(torch.tensor([1000.0, 1001.0, 1002.0])), torch.softmax(torch.tensor([0.0, 1.0, 2.0]), dim=0)))
print("все проверки пройдены")
```
