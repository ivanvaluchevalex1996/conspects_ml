# Deep-ML #1218 · Count Parameters of a Sequential Model · **easy**

Посчитай число обучаемых параметров модели `nn.Module` (в том числе `nn.Sequential`): сумма `numel()` по `model.parameters()` с `requires_grad=True`.

[#1218](https://www.deep-ml.com/problems/1218) · [README.md](./README.md) · эталон [1218_count_parameters_sequential_решения.ipynb](./1218_count_parameters_sequential_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def count_params(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)
```

```python
need(count_params(nn.Sequential(nn.Linear(4, 3), nn.ReLU(), nn.Linear(3, 2))) == 23)
print("все проверки пройдены")
```
