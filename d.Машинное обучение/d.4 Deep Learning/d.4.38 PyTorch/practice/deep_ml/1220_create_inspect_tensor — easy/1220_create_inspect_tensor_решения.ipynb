# Deep-ML #1220 · Create and Inspect a Tensor · **easy**

Создай тензор `float32` из вложенного списка и верни кортеж `(shape, dtype, ndim)`. `dtype` — строка без префикса `torch`, например `"float32"`.

[#1220](https://www.deep-ml.com/problems/1220) · [README.md](./README.md) · эталон [1220_create_inspect_tensor_решения.ipynb](./1220_create_inspect_tensor_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def inspect_tensor(data):
    t = torch.tensor(data, dtype=torch.float32)
    return tuple(t.shape), str(t.dtype).replace("torch.", ""), t.dim()
```

```python
need(inspect_tensor([[1, 2, 3], [4, 5, 6]]) == ((2, 3), "float32", 2))
need(inspect_tensor([7]) == ((1,), "float32", 1))
print("все проверки пройдены")
```
