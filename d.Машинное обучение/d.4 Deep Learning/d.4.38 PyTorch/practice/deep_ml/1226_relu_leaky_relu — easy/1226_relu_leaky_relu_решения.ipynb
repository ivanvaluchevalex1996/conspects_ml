# Deep-ML #1226 · Implement ReLU and Leaky ReLU · **easy**

Реализуй `relu(x)` и `leaky_relu(x, alpha=0.01)` для тензора `x` без `F.relu` и `F.leaky_relu`: используй `torch.where` или арифметику.

[#1226](https://www.deep-ml.com/problems/1226) · [README.md](./README.md) · эталон [1226_relu_leaky_relu_решения.ipynb](./1226_relu_leaky_relu_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def relu(x):
    return torch.where(x > 0, x, torch.zeros_like(x))


def leaky_relu(x, alpha=0.01):
    return torch.where(x > 0, x, alpha * x)
```

```python
need(close(relu(torch.tensor([-2.0, 0.0, 3.0])), [0.0, 0.0, 3.0]))
need(close(leaky_relu(torch.tensor([-2.0, 0.0, 3.0]), 0.1), [-0.2, 0.0, 3.0]))
print("все проверки пройдены")
```
