# Deep-ML #1225 · Two-Layer MLP Forward Pass · **medium**

Напиши прямой проход двухслойного MLP: `x` (N×in), `W1` (hidden×in), `b1` (hidden), `W2` (out×hidden), `b2` (out). Скрытый слой — `ReLU(x·W1ᵀ + b1)`, выход — линейный: `h·W2ᵀ + b2`.

[#1225](https://www.deep-ml.com/problems/1225) · [README.md](./README.md) · эталон [1225_mlp_two_layer_forward_решения.ipynb](./1225_mlp_two_layer_forward_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def mlp_forward(x, W1, b1, W2, b2):
    h = F.relu(x @ W1.T + b1)
    return h @ W2.T + b2
    
```

```python
x = torch.tensor([[1.0, 2.0, 3.0]])
W1 = torch.tensor([[1.0, 0.0, 2.0], [0.0, -1.0, 3.0]])
b1 = torch.zeros(2)
W2 = torch.tensor([[2.0, 3.0]])
b2 = torch.tensor([1.0])

print(mlp_forward(x, W1, b1, W2, b2))
```

```python
need(close(mlp_forward(torch.tensor([[1.0, 2.0]]), torch.tensor([[1.0, 0.0], [0.0, -1.0]]), torch.zeros(2), torch.tensor([[2.0, 3.0]]), torch.tensor([1.0])), [[3.0]]))


mlp_forward(torch.tensor([[1.0, 2.0]]), torch.tensor([[1.0, 0.0], [0.0, -1.0]]), torch.zeros(2), torch.tensor([[2.0, 3.0]]), torch.tensor([1.0])), [[3.0]]
```
