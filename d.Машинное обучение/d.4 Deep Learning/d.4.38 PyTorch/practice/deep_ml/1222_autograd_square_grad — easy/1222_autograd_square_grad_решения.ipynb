# Deep-ML #1222 · Gradient of a Square with Autograd · **easy**

Для функции f(x) = x² найди производную df/dx в точке `x` через `autograd`. Верни число.

[#1222](https://www.deep-ml.com/problems/1222) · [README.md](./README.md) · эталон [1222_autograd_square_grad_решения.ipynb](./1222_autograd_square_grad_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def square_grad(x):
    x_t = torch.tensor(float(x), requires_grad=True)
    y = x_t ** 2
    y.backward()
    return x_t.grad.item()
```

```python
need(close(square_grad(3.0), 6.0))
need(close(square_grad(-1.5), -3.0))
print("все проверки пройдены")
```
