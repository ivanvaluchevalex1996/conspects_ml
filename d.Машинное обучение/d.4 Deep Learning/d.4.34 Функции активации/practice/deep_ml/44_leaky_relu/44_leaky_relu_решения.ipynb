```python
import numpy as np
import torch
import torch.nn.functional as F


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"
```

```python
def leaky_relu(z: float, alpha: float = 0.01) -> float:
    return z if z > 0 else alpha * z


need(leaky_relu(-1) == -0.01)
need(abs(leaky_relu(-2, alpha=0.1) - (-0.2)) < 1e-9)
```

```python
def leaky_relu_numpy(x: np.ndarray, alpha: float) -> np.ndarray:
    return np.where(x > 0, x, alpha * x)


z_t = torch.tensor([-2.0, -0.5, 1.0, 3.0])
alpha = 0.1
mine = torch.tensor(leaky_relu_numpy(z_t.numpy(), alpha))
need(torch.allclose(mine, F.leaky_relu(z_t, alpha)))
print(mine.round(decimals=2).tolist())
```
