````python
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"
````

````python
def relu(z: float) -> float:
    return z if z > 0 else 0.0


def relu_derivative(z: float) -> float:
    return 1.0 if z > 0 else 0.0  # z=0 как Deep-ML #42


need(relu(0.0) == 0.0 and relu(1.0) == 1.0 and relu(-1.0) == 0.0)
````

````python
def relu_numpy(x: np.ndarray) -> np.ndarray:
    return np.maximum(x, 0.0)


x_np = np.array([-1.0, 0.0, 2.5])
need(np.allclose(relu_numpy(x_np), F.relu(torch.tensor(x_np)).numpy()))
````

````python
xs = np.linspace(-3, 3, 100)
plt.figure(figsize=(6, 3))
plt.plot(xs, [relu(float(t)) for t in xs], label="ReLU")
plt.plot(xs, [relu_derivative(float(t)) for t in xs], label="ReLU′")
plt.legend()
plt.grid(True, alpha=0.3)
plt.title("ReLU и производная")
plt.show()
````
