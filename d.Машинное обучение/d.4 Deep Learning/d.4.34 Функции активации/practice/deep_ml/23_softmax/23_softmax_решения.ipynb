# Пример · #23 Softmax (эталон)

После [23_softmax.ipynb](./23_softmax.ipynb).

1. Стабильный softmax (вычесть max).
2. Сверка с `F.softmax(..., dim=0)`.
3. Большие логиты — без nan.


```python
import numpy as np
import torch
import torch.nn.functional as F


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"
```

## Уровень 1 · NumPy

```python
def softmax(scores: list[float]) -> list[float]:
    arr = np.array(scores, dtype=float)
    shifted = arr - np.max(arr)
    exp_s = np.exp(shifted)
    probs = exp_s / exp_s.sum()
    return probs.tolist()


_p = softmax([1, 2, 3])
need(abs(sum(_p) - 1.0) < 1e-6)
need([round(x, 4) for x in _p] == [0.0900, 0.2447, 0.6652])
print("softmax([1,2,3]) =", [round(x, 4) for x in _p])
```

## Уровень 2 · PyTorch + стабильность

```python
z = torch.tensor([1.0, 2.0, 3.0])
mine = np.array(softmax(z.tolist()), dtype=float)
ref = F.softmax(z, dim=0).numpy()
need(np.allclose(mine, ref, rtol=1e-5))
print("torch:", ref.round(4))

big = np.array([1000.0, 1001.0, 1002.0])
p_big = softmax(big.tolist())
need(abs(sum(p_big) - 1.0) < 1e-6 and not np.isnan(p_big).any())
print("big logits:", np.round(p_big, 3))
```
