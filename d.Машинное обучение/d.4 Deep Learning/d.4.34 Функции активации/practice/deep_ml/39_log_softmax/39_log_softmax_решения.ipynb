```python
import numpy as np
import torch
import torch.nn.functional as F


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"
```

```python
def log_softmax(scores: np.ndarray) -> np.ndarray:
    shifted = scores - np.max(scores)
    return shifted - np.log(np.sum(np.exp(shifted)))


need(
    np.allclose(
        np.round(log_softmax(np.array([1.0, 2.0, 3.0])), 4),
        [-2.4076, -1.4076, -0.4076],
    )
)
print("log_softmax:", np.round(log_softmax(np.array([1.0, 2.0, 3.0])), 4))
```

```python
z = torch.tensor([1.0, 2.0, 3.0])
mine = log_softmax(z.numpy())
ref = F.log_softmax(z, dim=0).numpy()
need(np.allclose(mine, ref, rtol=1e-5))
print("mine:", mine.round(4), "torch:", ref.round(4))
```
