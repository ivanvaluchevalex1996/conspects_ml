# Пример · #1235 momentum_step (эталон) · **easy** / **medium**

После [1235_momentum_step.ipynb](./1235_momentum_step.ipynb).

```python
import torch


def momentum_step(w, grad, v, lr, mu):
    v = mu * v + grad
    w = w - lr * v
    return w, v


w_new, v_new = momentum_step(
    torch.tensor([1.0, 2.0]),
    torch.tensor([0.1, 0.2]),
    torch.tensor([0.0, 0.0]),
    0.1,
    0.9,
)
assert torch.allclose(w_new, torch.tensor([0.9900, 1.9800]))
assert torch.allclose(v_new, torch.tensor([0.1000, 0.2000]))
print(w_new, v_new)
```
