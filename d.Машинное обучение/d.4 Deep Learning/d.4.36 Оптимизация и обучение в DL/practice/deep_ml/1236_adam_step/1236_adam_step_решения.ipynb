# Пример · #1236 adam_step (эталон)

После [1236_adam_step.ipynb](./1236_adam_step.ipynb).

```python
import torch


def adam_step(w, grad, m, v, t, lr, beta1, beta2, eps):
    m = beta1 * m + (1 - beta1) * grad
    v = beta2 * v + (1 - beta2) * grad ** 2
    m_hat = m / (1 - beta1**t)
    v_hat = v / (1 - beta2**t)
    w = w - lr * m_hat / (torch.sqrt(v_hat) + eps)
    return w, m, v


w_new, m_new, v_new = adam_step(
    torch.tensor([1.0, 2.0]),
    torch.tensor([0.5, -0.5]),
    torch.zeros(2),
    torch.zeros(2),
    1,
    0.001,
    0.9,
    0.999,
    1e-8,
)
assert torch.allclose(w_new, torch.tensor([0.9990, 2.0010]), atol=1e-4)
assert torch.allclose(m_new, torch.tensor([0.0500, -0.0500]), atol=1e-4)
assert torch.allclose(v_new, torch.tensor([0.0003, 0.0003]), atol=1e-4)
print(w_new, m_new, v_new)
```
