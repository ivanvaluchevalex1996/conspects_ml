## Зачем эта задача

SGD — базовый шаг: **в минус градиент**. Все остальные оптимизаторы — надстройки.


## Условие

`sgd_step(w, grad, lr)` → `w - lr*grad`.


```python
import torch


def sgd_step(w, grad, lr):
    return w - lr * grad


out = sgd_step(torch.tensor([1.0, 2.0]), torch.tensor([0.5, 1.0]), 0.1)
assert torch.allclose(out, torch.tensor([0.9500, 1.9000]))
print(out)
```

## Построчно

- Поэлементно на тензорах.
- assert проверяет один шаг с lr=0.1.
