# Deep-ML #146 · momentum_optimizer · **easy**

**Уровень:** easy (Deep-ML); доп. к handbook.

[#146](https://www.deep-ml.com/problems/146) · [README.md](./README.md) · эталон [146_momentum_решения.ipynb](./146_momentum_решения.ipynb) **после** попытки

```python
def momentum_optimizer(parameter, grad, velocity, learning_rate=0.01, momentum=0.9):
    velocity = momentum * velocity + learning_rate * grad
    parameter = parameter - velocity
    return parameter, velocity


p, v = momentum_optimizer(1.0, 0.1, 0.1)
assert round(p, 3) == 0.909 and round(v, 3) == 0.091
print(round(p, 3), round(v, 3))
```
