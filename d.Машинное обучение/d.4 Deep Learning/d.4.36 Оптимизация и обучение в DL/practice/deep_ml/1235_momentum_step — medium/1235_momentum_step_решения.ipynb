## Зачем эта задача

**Импульс** копит скорость v = μv + grad; шаг по v сглаживает осцилляции в овражках.


## Условие

Обновить w и v; эталон из assert.


```python
import torch


def momentum_step(w, grad, v, lr, mu):
    v = mu * v + grad
    w = w - lr * v
    return w, v


w_new, v_new = momentum_step(
    torch.tensor([1.0, 2.0]),
    torch.tensor([0.1, 0.2]),
    torch.tensor([0.0, 0.0]),
    0.1,
    0.9,
)
assert torch.allclose(w_new, torch.tensor([0.9900, 1.9800]))
assert torch.allclose(v_new, torch.tensor([0.1000, 0.2000]))
print(w_new, v_new)
```

## Построчно

- Сначала v, потом `w -= lr*v` (как в задаче).
- Отличие от #146: там lr внутри velocity — смотри условие каждой задачи.
