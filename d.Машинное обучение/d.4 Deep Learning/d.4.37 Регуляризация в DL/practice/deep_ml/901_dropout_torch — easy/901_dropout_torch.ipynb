# Deep-ML #901 · dropout (torch) · **easy**

[#901](https://www.deep-ml.com/problems/901) · [README.md](./README.md) · эталон [901_dropout_torch_решения.ipynb](./901_dropout_torch_решения.ipynb) **после** попытки

```python
import torch


def dropout(x, p, training):
    if not training:
        return x
    mask = (torch.rand_like(x) > p).float()
    return x * mask / (1.0 - p)


x = torch.ones(1000)
y = dropout(x, 0.5, training=True)
assert torch.allclose(x.mean(), y.mean(), atol=0.15)
assert torch.allclose(dropout(x, 0.5, training=False), x)
```
