# Пример · #1231 bn_eval (эталон) · **medium**

После [1231_bn_eval.ipynb](./1231_bn_eval.ipynb).

```python
import torch


def bn_eval(x, mean, var, gamma, beta, eps=1e-5):
    return (x - mean) / torch.sqrt(var + eps) * gamma + beta


out = bn_eval(
    torch.tensor([1.0, 2.0, 3.0]),
    torch.tensor(2.0),
    torch.tensor(1.0),
    torch.tensor(1.0),
    torch.tensor(0.0),
)
assert torch.allclose(out, torch.tensor([-1.0, 0.0, 1.0]), atol=1e-4)
print(out)
```
