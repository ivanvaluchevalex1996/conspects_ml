## Зачем эта задача

Формула BatchNorm **в eval**: фиксированные mean/var (running stats), γ и β.


## Условие

`bn_eval` на трёх точках → [-1,0,1] при заданных статистиках.


```python
import torch


def bn_eval(x, mean, var, gamma, beta, eps=1e-5):
    return (x - mean) / torch.sqrt(var + eps) * gamma + beta


out = bn_eval(
    torch.tensor([1.0, 2.0, 3.0]),
    torch.tensor(2.0),
    torch.tensor(1.0),
    torch.tensor(1.0),
    torch.tensor(0.0),
)
assert torch.allclose(out, torch.tensor([-1.0, 0.0, 1.0]), atol=1e-4)
print(out)
```

## Построчно

- Нормализация (x-mean)/√(var+ε).
- Аффин γ·x̂+β.
