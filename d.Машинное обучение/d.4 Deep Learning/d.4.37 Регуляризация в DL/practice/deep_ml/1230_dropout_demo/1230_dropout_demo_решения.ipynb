# Пример · #1230 dropout_demo (эталон)

После [1230_dropout_demo.ipynb](./1230_dropout_demo.ipynb).

```python
import torch
import torch.nn as nn


def dropout_demo():
    torch.manual_seed(0)
    x = torch.ones(10)
    layer = nn.Dropout(0.5)
    layer.train()
    _ = layer(x)
    nz = int((layer(x) != 0).sum().item())
    layer.eval()
    eval_out = layer(x)
    return eval_out, nz


eval_out, nz = dropout_demo()
assert torch.allclose(eval_out, torch.ones(10))
assert nz == 4
print(eval_out, nz)
```
