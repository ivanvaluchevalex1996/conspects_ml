```python
import numpy as np


class DropoutLayer:
    def __init__(self, p: float):
        self.p = p
        self.mask = None

    def forward(self, x: np.ndarray, training: bool = True) -> np.ndarray:
        if not training:
            return x
        self.mask = (np.random.rand(*x.shape) > self.p).astype(float)
        return x * self.mask / (1.0 - self.p)

    def backward(self, grad: np.ndarray) -> np.ndarray:
        return grad * self.mask / (1.0 - self.p)


np.random.seed(42)
x = np.array([1.0, 2.0, 3.0, 4.0])
grad = np.array([0.1, 0.2, 0.3, 0.4])
layer = DropoutLayer(0.5)
assert np.allclose(layer.forward(x, training=True), [0.0, 4.0, 6.0, 8.0])
assert np.allclose(layer.backward(grad), [0.0, 0.4, 0.6, 0.8])
print("ok")
```
