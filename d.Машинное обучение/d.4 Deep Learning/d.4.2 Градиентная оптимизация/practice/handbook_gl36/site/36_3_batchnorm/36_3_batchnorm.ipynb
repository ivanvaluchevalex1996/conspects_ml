# §36.3 · BatchNorm1d

`Linear → BatchNorm → ReLU`, батч 32×10 → выход 32×20.

```python
import torch
from torch import nn

layer = nn.Sequential(nn.Linear(10, 20), nn.BatchNorm1d(20), nn.ReLU())
x = torch.randn(32, 10)
y = layer(x)
assert tuple(y.shape) == (32, 20)
print("форма выхода:", tuple(y.shape))
```
