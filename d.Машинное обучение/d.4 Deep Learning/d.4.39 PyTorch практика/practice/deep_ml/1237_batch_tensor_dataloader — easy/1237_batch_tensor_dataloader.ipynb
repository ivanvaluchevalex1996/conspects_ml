# Deep-ML #1237 · Batch a TensorDataset with DataLoader · **easy**

Напиши `batch_stats(X, y)`: `TensorDataset`, `DataLoader(batch_size=4, shuffle=False)`, верни `(num_batches, first_batch_X_shape_tuple)` — кортеж Python, не `torch.Size`.

[https://www.deep-ml.com/problems/1237](https://www.deep-ml.com/problems/1237) · [README.md](./README.md) · эталон [1237_batch_tensor_dataloader_решения.ipynb](./1237_batch_tensor_dataloader_решения.ipynb) **после** попытки


```python
import torch
import torch.nn as nn
import torch.nn.functional as F


def need(cond, msg="допиши код"):
    assert cond, msg


def close(a, b, eps=1e-5):
    if a is None:
        return False
    return torch.allclose(torch.as_tensor(a, dtype=torch.float32),
                          torch.as_tensor(b, dtype=torch.float32), atol=eps)
```

```python
def batch_stats(X, y):
    from torch.utils.data import DataLoader, TensorDataset

    loader = DataLoader(TensorDataset(X, y), batch_size=4, shuffle=False)
    num_batches = 0
    first_shape = None
    for xb, _ in loader:
        if first_shape is None:
            first_shape = tuple(xb.shape)
        num_batches += 1
    return num_batches, first_shape
```

```python
X = torch.arange(16, dtype=torch.float32).reshape(8, 2)
y = torch.arange(8)
need(batch_stats(X, y) == (2, (4, 2)))
print("все проверки пройдены")
```
