## Домашка handbook · Deep-ML #1224

1. NumPy: `W = [[0.5, -0.2, 0.3]]`, `b = 0.1`, `x = [1, 2, 3]` — линейный слой **без** активации.
2. PyTorch: `nn.Sequential(nn.Linear(3, 1))`, те же веса под `torch.no_grad()`, тот же `x` — выход должен совпасть с NumPy.
3. Функция `single_neuron_forward(x)` как на [Deep-ML](https://www.deep-ml.com/problems/1224) (тензор `(1, 3)` → **float**).

Ожидание на примере Deep-ML: `single_neuron_forward(torch.tensor([[1., 2., 3.]]))` → сверь с условием на сайте (не подглядывай в решения).

```python
W = np.array([[0.5, -0.2, 0.3]])
b = 0.1
x_np = np.array([1.0, 2.0, 3.0])

#  одно число (float)
z_numpy = (W @ x_np + b).item()
need(isinstance(z_numpy, float))
print("NumPy z =", z_numpy)
```

```python
x_t = torch.tensor([[1.0, 2.0, 3.0]])
#  + задать weight/bias
net_lin = nn.Sequential(nn.Linear(3, 1))

with torch.no_grad():
    net_lin[0].weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
    net_lin[0].bias.copy_(torch.tensor([0.1]))


z_torch = net_lin(x_t).item()
need(abs(z_torch - z_numpy) < 1e-5, "PyTorch и NumPy должны совпасть")
print("PyTorch z =", z_torch)
```

```python
def single_neuron_forward(x):
    net = nn.Sequential(
        nn.Linear(3, 1),
    )
    with torch.no_grad():
        net[0].weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
        net[0].bias.copy_(torch.tensor([0.1]))
    return net(x).item()


out = single_neuron_forward(torch.tensor([[1.0, 2.0, 3.0]]))
need(isinstance(out, float))
need(abs(out - z_numpy) < 1e-5, "должно совпасть с NumPy из ячейки выше")
print("single_neuron_forward =", out)
```
