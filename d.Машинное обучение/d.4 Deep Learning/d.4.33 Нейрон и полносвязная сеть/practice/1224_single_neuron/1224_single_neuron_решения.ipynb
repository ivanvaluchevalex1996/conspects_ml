## Зачем эта задача

Один нейрон = `Linear` + веса; сверить **NumPy dot** и `nn.Linear` с теми же W,b.


## Условие

`single_neuron_forward(x)` → скаляр; фиксированные веса как в тесте.


```python
import numpy as np
import torch
import torch.nn as nn


def need(cond, msg=""):
    assert cond, msg or "проверка не прошла"
```

```python
W = np.array([[0.5, -0.2, 0.3]])
b = 0.1
x_np = np.array([1.0, 2.0, 3.0])
z_numpy = float((W @ x_np + b).item())
need(isinstance(z_numpy, float))
print("NumPy z =", z_numpy)
```

```python
x_t = torch.tensor([[1.0, 2.0, 3.0]])
net_lin = nn.Sequential(nn.Linear(3, 1))
with torch.no_grad():
    net_lin[0].weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
    net_lin[0].bias.copy_(torch.tensor([0.1]))
z_torch = net_lin(x_t).item()
need(abs(z_torch - z_numpy) < 1e-5)
print("PyTorch z =", z_torch)
```

```python
def single_neuron_forward(x):
    net = nn.Sequential(nn.Linear(3, 1))
    with torch.no_grad():
        net[0].weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
        net[0].bias.copy_(torch.tensor([0.1]))
    return net(x).item()


out = single_neuron_forward(torch.tensor([[1.0, 2.0, 3.0]]))
need(isinstance(out, float) and abs(out - z_numpy) < 1e-5)
print("single_neuron_forward =", out)
```

## Построчно

- NumPy: `W @ x + b`.
- PyTorch: копируем веса в `nn.Linear(3,1)` под `no_grad`.
- Функция для submit — тот же forward.
