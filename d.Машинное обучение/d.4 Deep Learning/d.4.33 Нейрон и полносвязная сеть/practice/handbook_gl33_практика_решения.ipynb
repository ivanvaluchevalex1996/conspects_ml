## §33.1

## Зачем этот фрагмент

Формула одного нейрона: **линейная комбинация** → **активация** φ. Так же устроен один `nn.Linear` + активация в PyTorch. См. [d.4.33 …](../d.4.33%20Нейрон%20и%20полносвязная%20сеть/d.4.33%20Нейрон%20и%20полносвязная%20сеть.md) §1.


```python
import numpy as np


def neuron(x, w, b, phi):
    return phi(w @ x + b)


sigmoid = lambda z: 1 / (1 + np.exp(-z))
x = np.array([0.5, -1.2, 2.0])
w = np.array([0.4, 0.1, -0.3])
b = 0.2
print("выход нейрона:", round(neuron(x, w, b, sigmoid), 4))
```

## §33.4

## Зачем этот фрагмент

`nn.Sequential` — минимальная **полносвязная** сеть: считаем число параметров и один forward. Это мост к handbook и к project-моделям.


```python
import torch
import torch.nn as nn

net = nn.Sequential(
    nn.Linear(3, 4),
    nn.ReLU(),
    nn.Linear(4, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid(),
)
x = torch.tensor([[0.5, -1.2, 2.0]])
print("выход сети:", net(x).item())
print("параметров:", sum(p.numel() for p in net.parameters()))
```

## §33.5

## Зачем этот фрагмент

Ручной forward **двух слоёв** в NumPy: видно `z = Wx+b`, затем σ, затем второй слой. Так проверяют понимание до autograd.


```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


x = np.array([1.0, 0.5])
W1 = np.array([[0.3, -0.1], [0.2, 0.4]])
b1 = np.array([0.0, 0.1])
W2 = np.array([0.7, -0.5])
b2 = 0.2

z1 = W1 @ x + b1
a1 = sigmoid(z1)
z2 = W2 @ a1 + b2
a2 = sigmoid(z2)
print("z1 =", z1.round(3), "a1 =", a1.round(3))
print("z2 =", round(float(z2), 3), "a2 =", round(float(a2), 3))
```

## Домашка · NumPy, PyTorch, Deep-ML #1224

## Зачем этот фрагмент

Сверка **NumPy / PyTorch / Deep-ML #1224**: один и тот же скалярный выход `z` тремя способами — эталон для submit.


```python
import numpy as np
import torch
import torch.nn as nn

W = np.array([[0.5, -0.2, 0.3]])
b = 0.1
x_np = np.array([1.0, 2.0, 3.0])
z_numpy = (W @ x_np + b).squeeze().item()
print("NumPy", z_numpy)

x_t = torch.tensor([[1.0, 2.0, 3.0]])
net_lin = nn.Sequential(nn.Linear(3, 1))
with torch.no_grad():
    net_lin[0].weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
    net_lin[0].bias.copy_(torch.tensor([0.1]))
print("PyTorch", net_lin(x_t).item())


def single_neuron_forward(x):
    layer = nn.Linear(3, 1)
    with torch.no_grad():
        layer.weight.copy_(torch.tensor([[0.5, -0.2, 0.3]]))
        layer.bias.copy_(torch.tensor([0.1]))
    return layer(x).item()


print("Deep-ML example", single_neuron_forward(x_t))
```
