## §33.1 — один нейрон

По [§33.1 handbook](https://mlhandbook-production.up.railway.app/parts/deep-learning/neural-network.html): `neuron(x, w, b, phi) = phi(w @ x + b)`, sigmoid, данные как на сайте.

Сверь с handbook: `выход нейрона: 0.4207`

```python
def neuron(x, w, b, phi):
    return phi(w @ x + b)


sigmoid = lambda z: 1 / (1 + np.exp(-z))
x = np.array([0.5, -1.2, 2.0])
w = np.array([0.4, 0.1, -0.3])
b = 0.2

out = neuron(x, w, b, sigmoid)
need(isinstance(out, (float, np.floating)))
print("выход нейрона:", round(float(out), 4))
```

Конспект: [`d.4.33`](../d.4.33%20%D0%9D%D0%B5%D0%B9%D1%80%D0%BE%D0%BD%20%D0%B8%20%D0%BF%D0%BE%D0%BB%D0%BD%D0%BE%D1%81%D0%B2%D1%8F%D0%B7%D0%BD%D0%B0%D1%8F%20%D1%81%D0%B5%D1%82%D1%8C.md) · чеклист: [`handbook_gl33_практика.ipynb`](./handbook_gl33_практика.ipynb)

## §33.4 — `Sequential` 3→4→4→1

Собери сеть как в ячейке handbook (Linear → ReLU → … → Sigmoid), `x = [[0.5, -1.2, 2.0]]`.

Сверь: **41** обучаемый параметр; вывод `net(x)` — как на сайте (число будет своё из‑за случайных весов, но **41** должно совпасть).

```python
net = nn.Sequential(
    nn.Linear(3, 4), nn.ReLU(),     # вход 3 → скрытый 4 + нелинейность
    nn.Linear(4, 4), nn.ReLU(),
    nn.Linear(4, 1), nn.Sigmoid(),  # выход — вероятность
)

x = torch.tensor([[0.5, -1.2, 2.0]])
n_params = sum(p.numel() for p in net.parameters())
need(n_params == 41, f"ожидали 41 параметр, получили {n_params}")
print("выход сети:", net(x).item())
print("всего обучаемых параметров:", n_params)
```

## §33.5 — forward 2→2→1 руками

Веса и вход — как в [§33.5](https://mlhandbook-production.up.railway.app/parts/deep-learning/neural-network.html#forward-%D0%BF%D1%80%D0%BE%D1%85%D0%BE%D0%B4-%D1%80%D1%83%D0%BA%D0%B0%D0%BC%D0%B8-%D0%BF%D0%BE-%D1%88%D0%B0%D0%B3%D0%B0%D0%BC). Посчитай `z1, a1, z2, a2` (sigmoid между слоями).

Сверь с handbook (округление): `z1 ≈ [0.25, 0.5]`, `a1 ≈ [0.562, 0.622]`, `z2 ≈ 0.282`, `a2 ≈ 0.57`.

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


x = np.array([1.0, 0.5])
W1 = np.array([[0.3, -0.1], [0.2, 0.4]])
b1 = np.array([0.0, 0.1])
W2 = np.array([0.7, -0.5])
b2 = 0.2

z1 = W1 @ x + b1; 
a1 = sigmoid(z1) 
z2 = W2 @ a1 + b2
a2 = sigmoid(z2)

need(z1 is not None and a2 is not None)
print("скрытый слой z1 =", np.round(z1, 3), "→ a1 =", np.round(a1, 3))
print(
    "выход        z2 =",
    round(float(z2), 3),
    "→ a2 =",
    round(float(a2), 3),
)
```
